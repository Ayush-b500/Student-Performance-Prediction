```python
import numpy as np
import pandas as pd

np.random.seed(42)

data = {
    "study_hours": np.random.randint(1, 10, 50),
    "attendance": np.random.randint(50, 100, 50),
    "previous_score": np.random.randint(40, 90, 50),
}

df = pd.DataFrame(data)

# target column
df["final_score"] = (
    df["study_hours"] * 5 +
    df["attendance"] * 0.3 +
    df["previous_score"] * 0.5
)

df.head()
```
